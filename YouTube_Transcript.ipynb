```python
import os
import json
from youtube_transcript_api import YouTubeTranscriptApi


input_json_path = "youtube-data-id.json"
output_folder = "transcripts"


os.makedirs(output_folder, exist_ok=True)


with open(input_json_path, "r", encoding="utf-8") as f:
    data = json.load(f)

channel_name = data.get("channel", "Unknown Channel")
videos = data.get("videos", [])

print(f"Starting transcript downloads for channel: {channel_name}\n" + "-"*50)


for video in videos:
    title = video["title"]
    video_id = video["id"]
    

    output_file_path = os.path.join(output_folder, f"{video_id}.json")
    
    print(f"Processing: '{title}' ({video_id})...")
    
    try:

        transcript_data = YouTubeTranscriptApi.get_transcript(video_id)
        
   
        with open(output_file_path, "w", encoding="utf-8") as out_f:
            json.dump(transcript_data, out_f, ensure_ascii=False, indent=4)
            
        print(f" Saved -> {output_file_path}")
        
    except Exception as e:
        print(f"Failed to get transcript: {e}")

print("-"*50 + "\nAll processes completed!")
```
